```python
def least_squares_second():
  # Load the diabetes dataset
  diabetes_X, diabetes_y = datasets.load_diabetes(return_X_y=True)

  # only the second feature
  diabetes_X = diabetes_X[:, 1]
  diabetes_X_train = diabetes_X[:-20].reshape(-1, 1)
  diabetes_X_test = diabetes_X[-20:].reshape(-1, 1)
  diabetes_y_train = diabetes_y[:-20]
  diabetes_y_test = diabetes_y[-20:]


  A = np.hstack([np.ones((diabetes_X_train.shape[0], 1)), diabetes_X_train])
  X_test = np.hstack([np.ones((diabetes_X_test.shape[0], 1)), diabetes_X_test])

  # compute coefficients using matrix method
  alpha = np.dot((np.dot(np.linalg.inv(np.dot(A.T, A)), A.T)), diabetes_y_train)

  # predict target values
  diabetes_y_pred = np.dot(X_test, alpha)

  # compute the mean squared error, r2
  
  mse = np.mean((diabetes_y_test - diabetes_y_pred)**2)
  r2 = r2_score(diabetes_y_test, diabetes_y_pred)
  return mse, r2


# record runtime
start = time.time()
least_squares_second()
end = time.time()
elapsed = end - start

mse, r2 = least_squares_second()

print(f'MSE: {mse:.3f}')
print(f'r2: {r2:.3f}')
print(f'Runtime: {elapsed:.5f}s')
```

```python
def gradient_descent():
  # Load the diabetes dataset
  diabetes_X, diabetes_y = datasets.load_diabetes(return_X_y=True)

  # first two features
  diabetes_X = diabetes_X[:, :2]

  # split the data into training and testing sets
  diabetes_X_train = diabetes_X[:-20]
  diabetes_X_test = diabetes_X[-20:]
  diabetes_y_train = diabetes_y[:-20]
  diabetes_y_test = diabetes_y[-20:]

  diabetes_X_train = np.hstack((np.ones((diabetes_X_train.shape[0], 1)), diabetes_X_train))
  diabetes_X_test = np.hstack((np.ones((diabetes_X_test.shape[0], 1)), diabetes_X_test))

  # learning rate and initialized theta to zeros
  learning_rate = 0.01
  theta = np.zeros(diabetes_X_train.shape[1])

  # gradient descent using matrix loss equation
  N = diabetes_X_train.shape[0]
  for i in range(1000):
      diabetes_y_pred = np.dot(diabetes_X_train, theta)
      error = diabetes_y_pred - diabetes_y_train
      gradient_loss = (np.dot(diabetes_X_train.T, error) / (2 * N))
      theta = theta - learning_rate * gradient_loss

  # prediction using test set
  diabetes_y_pred = np.dot(diabetes_X_test, theta)

  # mean squared error
  mse = np.mean((diabetes_y_pred - diabetes_y_test) ** 2)
  r2 = r2_score(diabetes_y_test, diabetes_y_pred)
  return mse, r2

# record runtime
start = time.time()
gradient_descent()
end = time.time()
elapsed = end - start

mse, r2 = gradient_descent()

print(f'MSE: {mse:.3f}')
print(f'r2: {r2:.3f}')
print(f'Runtime: {elapsed:.5f}s')
```

```python
from sklearn.datasets import load_breast_cancer

# load the breast cancer dataset
data = load_breast_cancer()
cancer_X = data.data[:, :2]  # extract the first two features
cancer_y = data.target

# set hyperparameters
lr = 0.01
epochs = 1000
# Initialize weights and bias to zeros
w = np.zeros(cancer_X.shape[1])
b = 0


# normalize features
def normalize(X):
    mean = np.mean(X, axis=0)
    std = np.std(X, axis=0)
    X = (X - mean) / std
    return X


# Define the sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# Define the loss function
def loss_function(W, b, X, y):
    z = np.dot(X, W) + b
    h = sigmoid(z)
    J = -(1 / len(X)) * (np.dot(y, np.log(h)) + np.dot((1 - y), np.log(1 - h)))
    return J


# Define the gradient of the loss function
def gradient(w, b, X, y, h):
    dW = (1 / len(X)) * np.dot(X.T, (h - y))
    db = (1 / len(X)) * np.sum(h - y)
    return dW, db


def train(X, y, w, b, epochs, lr):
    # normalize features
    X = normalize(X)
    # Perform gradient descent to optimize weights and bias
    for i in range(epochs):
        h = sigmoid(np.dot(X, w) + b)
        dW, db = gradient(w, b, X, y, h)
        w -= lr * dW
        b -= lr * db

    return w, b


# Make predictions
def predict(X):
    X = normalize(X)
    pred = sigmoid(np.dot(X, w) + b)
    pred_class = np.where(pred >= 0.5, 1, 0)
    return pred_class


def accuracy(y, y_pred):
    accuracy = np.sum(predict(cancer_X) == y) / len(y)
    return accuracy


start = time.time()
# train
w, b = train(cancer_X, cancer_y, w, b, epochs, lr)
end = time.time()

elapsed = end - start

# Compute and print accuracy
print(f"Accuracy: {accuracy(cancer_y, predict(cancer_X)):.3f}")
# Print runtime
print(f"Runtime: {elapsed:.5f}s")
```
