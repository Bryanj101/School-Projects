```python
!pip install scikit-optimize
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from skopt import gp_minimize
from skopt.space import Real, Integer
from skopt.utils import use_named_args

# Load the breast cancer dataset
data = load_breast_cancer()

# Define the hyperparameters and their search space
space = [
    Integer(1, 10, name='max_depth'),
    Integer(2, 50, name='n_estimators'),
]

# Define the objective function to be optimized
@use_named_args(space)
def objective(**params):
    clf = RandomForestClassifier(**params)
    return -cross_val_score(clf, data.data, data.target, cv=5, n_jobs=-1).mean()

# Run the Bayesian optimization algorithm
result = gp_minimize(
    objective,
    space,
    n_calls=10,
    random_state=42,
)

# Print the optimal hyperparameters and function value
print(result.x)
print(result.fun)
```
